```python

import pandas as pd

dados = {
    "Produto": ["Notebook", "Mouse", "Teclado", "Monitor", "Headset"],
    "Quantidade": [5, 20, 12, 8, 15],
    "Preco_Unitario": [3500, 80, 150, 900, 200]
}

df = pd.DataFrame(dados)

df["Faturamento"] = df["Quantidade"] * df["Preco_Unitario"]

display(df)
```

```python

import matplotlib.pyplot as plt

plt.figure(figsize=(9, 5))

plt.bar(df["Produto"], df["Faturamento"])

plt.title("Faturamento por Produto")
plt.xlabel("Produto")
plt.ylabel("Faturamento (R$)")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()
```

```python

total = df["Faturamento"].sum()
media = df["Faturamento"].mean()

mais_vendido = df.loc[df["Quantidade"].idxmax(), "Produto"]
maior_faturamento = df.loc[df["Faturamento"].idxmax(), "Produto"]

print(f"Faturamento total: R$ {total:,.2f}")
print(f"Faturamento médio por produto: R$ {media:,.2f}")
print(f"Produto com maior quantidade vendida: {mais_vendido}")
print(f"Produto com maior faturamento: {maior_faturamento}")
```
